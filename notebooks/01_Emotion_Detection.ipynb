# MoodCare - Emotion Detection Module

This notebook implements emotion detection using a Hugging Face Transformer model.
The detected emotion will later help MoodCare generate personalized responses.

```python
from transformers import pipeline
```

```python
import random
```

```python
emotion_classifier = pipeline("text-classification",
                             model = "j-hartmann/emotion-english-distilroberta-base",
                             top_k = None)
```

```python
text = "I feel lonely and nobody wants to talk to me."

result = emotion_classifier(text)

result
```

```python
def detect_emotion(text):
    result = emotion_classifier(text)
    emotions = result[0]
    top_emotion = emotions[0]
    return {
        "emotion": top_emotion["label"],
        "confidence": round(top_emotion["score"] * 100, 2)
    }
```

```python
detect_emotion("I feel lonely and nobody wants to talk to me.")
```

```python
test_sentences = [
    "I am very happy today",
    "I feel lonely and ignored",
    "I am angry with everyone",
    "I am scared about my future",
    "Nothing interesting is happening, I am bored"
]

for sentence in test_sentences:
    print(sentence)
    print(detect_emotion(sentence))
    print("-------------------------------")
```

```python
emotion_responses = {
    "sadness": [
        "I understand that you are feeling low. Talking to someone you trust can help.",
        "You can try going for a short walk or doing something you enjoy.",
        "Would you like to share what is making you feel this way?"
    ],
    
    "anger": [
        "It seems like you are feeling frustrated. Taking a short break may help you calm down.",
        "Try deep breathing or stepping away from the situation for a while."
    ],
    
    "fear": [
        "It's completely normal to feel worried sometimes. Try focusing on things you can control.",
        "Talking about your worries with someone close can make things easier."
    ],
    
    "joy": [
        "That's great to hear! Keep enjoying this positive moment.",
        "You can share your happiness with someone you care about."
    ]
}
```

```python
def get_mood_response(emotion):
    if emotion in emotion_responses:
        return random.choice(emotion_responses[emotion])
    else:
        return "I am here to listen. Tell me more about how you are feeling."
```

```python
user_input = "I feel lonely and nobody understands me."

emotion_result = detect_emotion(user_input)

response = get_mood_response(emotion_result["emotion"])

print("Detected Emotion:", emotion_result['emotion'])
print("Confidence:", emotion_result['confidence'])
print("MoodCare:", response)
```
